## Limitations

- Some values are missing in the dataset.
- Some car prices are given as ranges instead of single values.
- Some seat values are written as ranges or formats like 2+2.
- Horsepower ranges are represented using their first number for analysis.
- The dataset may not include every car model available in the market.

## Conclusion

The exploratory data analysis of the Cars Datasets 2025 dataset helped identify patterns in car companies, fuel types, seating capacity, and horsepower.

Nissan had the highest number of entries in the dataset. Petrol was the most common fuel type, and five-seater cars were the most frequent. The average horsepower was approximately 301 hp. Horsepower and seating capacity showed a weak negative correlation.

The analysis also identified missing values, inconsistent fuel-type labels, price ranges, and unusual seating formats. These limitations should be considered while interpreting the results.

Overall, EDA helped summarize the dataset and understand its main characteristics through data cleaning, descriptive statistics, and visualizations.
